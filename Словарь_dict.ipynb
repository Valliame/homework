```python
%%writefile Словарь_dict.py
class Dict:
    def __init__(self, size=100):
        """Initialize the hash table."""
        self.size = size
        self.table = [[] for _ in range(size)]
        self.keys_list = []  # Store keys for iteration

    def _hash(self, key):
        """Hash function to determine index."""
        return hash(key) % self.size

    def __setitem__(self, key, value):
        """Add or update a value by key."""
        index = self._hash(key)
        for i, (k, v) in enumerate(self.table[index]):
            if k == key:
                self.table[index][i] = (key, value)
                return
        self.table[index].append((key, value))
        if key not in self.keys_list:
            self.keys_list.append(key)

    def __getitem__(self, key):
        """Retrieve a value by key."""
        index = self._hash(key)
        for k, v in self.table[index]:
            if k == key:
                return v
        raise KeyError(f"Key '{key}' not found")

    def __delitem__(self, key):
        """Delete a key and its value."""
        index = self._hash(key)
        for i, (k, _) in enumerate(self.table[index]):
            if k == key:
                del self.table[index][i]
                self.keys_list.remove(key)
                return
        raise KeyError(f"Key '{key}' not found")

    def __contains__(self, key):
        """Check if a key exists."""
        index = self._hash(key)
        return any(k == key for k, _ in self.table[index])

    def keys(self):
        """Retrieve all keys."""
        return self.keys_list

    def values(self):
        """Retrieve all values."""
        return [self[key] for key in self.keys_list]

    def items(self):
        """Retrieve all key-value pairs."""
        return [(key, self[key]) for key in self.keys_list]

    def __repr__(self):
        """String representation of the dictionary."""
        return "{" + ", ".join(f"'{key}': {repr(self[key])}" for key in self.keys_list) + "}"
```
