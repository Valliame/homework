```python
import pandas as pd

index = [
    ('city_1', 2010),
    ('city_1', 2020),
    ('city_2', 2010),
    ('city_2', 2020),
    ('city_3', 2010),
    ('city_3', 2020),
]

population = [
    101,
    201,
    102,
    202,
    103,
    203,
]

pop = pd.Series(population, index=pd.MultiIndex.from_tuples(index, names=["city", "year"]))

pop_df = pd.DataFrame(
    {
        'total': pop,
        'something': [10, 11, 12, 13, 14, 15]
    }
)

print(pop_df)
```

```python
pop_df_1 = pop_df.loc['city_1', 'something']
print(pop_df_1)
```

```python
pop_df_2 = pop_df.loc[['city_1', 'city_3'], ['total', 'something']]
print(pop_df_2)
```

```python
pop_df_3 = pop_df.loc[['city_1', 'city_3'], 'something']
print(pop_df_3)
```

```python
pop_df_2020 = pop_df.loc[pd.IndexSlice[:, 2020], :]
print(pop_df_2020)
```

```python
import numpy as np

# Мультииндекс для строк
index = pd.MultiIndex.from_product(
    [
        ['city_1', 'city_2'],
        [2010, 2020]
    ],
    names=['city', 'year']
)

# Мультииндекс для столбцов
columns = pd.MultiIndex.from_product(
    [
        ['person_1', 'person_2', 'person_3'],
        ['job_1', 'job_2']
    ],
    names=['worker', 'job']
)

data = np.arange(1, 25).reshape(4, 6)  
df = pd.DataFrame(data, index=index, columns=columns)
print(df)
```

```python
data_person1_3 = df.loc[:, pd.IndexSlice[['person_1', 'person_3'], :]]
print(data_person1_3)
```

```python
data_city1 = df.loc[
    'city_1',  
    pd.IndexSlice[['person_1', 'person_2'], :]  
]
print(data_city1)
```

```python
data_city1_slice = df.loc[
    'city_1',
    pd.IndexSlice['person_1':'person_2', :]  
]
print(data_city1_slice)
```

```python
ser1 = pd.Series(['a', 'b', 'c'], index=[1, 2, 3])
ser2 = pd.Series(['b', 'c', 'f'], index=[3, 4, 5])
```

```python
outer_join = pd.concat([ser1, ser2], join='outer')
print(outer_join)
```

```python
import pandas as pd

ser1 = pd.Series(['a', 'b', 'c'], index=[1, 2, 3])
ser2 = pd.Series(['b', 'c', 'f'], index=[3, 4, 5])

inner_join = pd.concat([ser1, ser2], join='inner', keys=['ser1', 'ser2'])
print(inner_join)
```

```python
import pandas as pd

ser1 = pd.Series(['a', 'b', 'c'], index=[1, 2, 3])
ser2 = pd.Series(['b', 'c', 'f'], index=[3, 4, 5])

common_index = ser1.index.intersection(ser2.index)

inner_join = pd.concat([ser1.loc[common_index], ser2.loc[common_index]], keys=['ser1', 'ser2'])

print(inner_join)
```
