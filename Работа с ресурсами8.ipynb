```python
import os

class ChangeDir:
    def __init__(self, path):
        self.new_path = path
        self.original_path = os.getcwd()

    def __enter__(self):
        if not os.path.exists(self.new_path):
            raise FileNotFoundError(f"Ошибка: Директория '{self.new_path}' не существует.")
        os.chdir(self.new_path)
        return self.new_path

    def __exit__(self, exc_type, exc_val, exc_tb):
        os.chdir(self.original_path)

# Example
print("Начальная директория:", os.getcwd())

try:
    with ChangeDir("/path/to/new/directory"):
        print("Внутри менеджера:", os.getcwd())
except FileNotFoundError as e:
    print(e)

print("После выхода:", os.getcwd())
```

```python
import sys

class RedirectOutput:
    def __init__(self, filename):
        self.filename = filename
        self.file = None
        self.original_stdout = sys.stdout

    def __enter__(self):
        self.file = open(self.filename, 'a', encoding='utf-8')  # Указание кодировки
        sys.stdout = self.file
        return self.file

    def __exit__(self, exc_type, exc_val, exc_tb):
        sys.stdout = self.original_stdout
        self.file.close()

# Пример 
print("Это стандартный вывод")

with RedirectOutput("output.txt"):
    print("Это вывод в файл")

print("Снова стандартный вывод")
```

```python
import time

class Timer:
    def __init__(self, name=""):
        self.name = name
        self.start_time = None

    def __enter__(self):
        self.start_time = time.time()

    def __exit__(self, exc_type, exc_val, exc_tb):
        elapsed_time = time.time() - self.start_time
        print(f"[{self.name}] Время выполнения: {elapsed_time:.3f} секунд")

# Example 
with Timer("Задача 1"):
    time.sleep(1)

with Timer("Задача 2"):
    for i in range(1000000):
        pass
```

```python
class SuppressExceptions:
    def __init__(self, *exceptions):
        self.exceptions = exceptions

    def __enter__(self):
        pass

    def __exit__(self, exc_type, exc_val, exc_tb):
        if exc_type and issubclass(exc_type, self.exceptions):
            print(f"Подавлено исключение: {exc_type.__name__}")
            return True 
        return False  

# Example 
with SuppressExceptions(ZeroDivisionError, ValueError):
    print(1 / 0)

with SuppressExceptions(TypeError):
    print(1 + "2")

with SuppressExceptions(IndexError):
    print([1, 2, 3][5])

print("Программа продолжает работать после блока with")
```

```python
import os
import tempfile

class TemporaryFile:
    def __enter__(self):
        self.temp_file = tempfile.NamedTemporaryFile(mode='w+', encoding='utf-8', delete=False)
        return self.temp_file

    def __exit__(self, exc_type, exc_val, exc_tb):
        try:
            self.temp_file.close()  
            os.remove(self.temp_file.name)
        except OSError as e:
            print(f"Ошибка при удалении файла: {e}")

# Пример 
with TemporaryFile() as temp_file:
    temp_file.write("Временные данные\n")  
    temp_file.seek(0)
    print(temp_file.read())

print("Файл автоматически удалён")
```
